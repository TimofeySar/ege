```python
a = '0123456789ABCDEF'
from itertools import product
import re
sh = 0
for q,w,e,r in product(a, repeat=4):
    s = q+w+e+r
    if q == '0':
        continue
    if s.count('E') == 1:
        s = re.sub('[02468AC]', '*', s)
        if s.count('*E') == 0 and s.count('E*') == 0:
            sh += 1
print(sh)
```

```python
f = [list(map(int, i.split())) for i in open('9.txt')]
sh = 0
for i in range(len(f)):
    a = list(set(f[i]))
    if len(a) == 4:
        if f[i].count(a[0]) == 4 or f[i].count(a[1]) == 4 or f[i].count(a[2]) == 4 or f[i].count(a[3]) == 4:
            if sorted(f[i], reverse=True) == f[i]:
                sh += i+1
print(sh)
```
