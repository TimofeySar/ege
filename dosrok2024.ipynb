```python
from itertools import product
sh = 1
m = 0
a = sorted('ПАРУС')
for i in product(a, repeat=5):
    s = ''.join(i)
    if s.count('У') <=1 and s.count('АА') == 0:
        m = sh
    sh += 1        
print(m)
```
